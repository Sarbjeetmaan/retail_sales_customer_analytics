```python
import pandas as pd

df = pd.read_excel("Online Retail.xlsx")
```

```python
print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print("Column names:", df.columns.tolist())
```

```python

# ==========================================
# STEP 2: DATA EXPLORATION
# ==========================================

# 1. Dataset Information
print("DATASET INFORMATION")
print("=" * 50)
df.info()


# 2. Missing Values
print("\nMISSING VALUES")
print("=" * 50)
print(df.isnull().sum())


# 3. Duplicate Records
print("\nDUPLICATE RECORDS")
print("=" * 50)
print("Total duplicates:", df.duplicated().sum())


# 4. Statistical Summary
print("\nSTATISTICAL SUMMARY")
print("=" * 50)
display(df.describe())


# 5. First 10 Records
print("\nFIRST 10 RECORDS")
print("=" * 50)
display(df.head(10))


# 6. Unique Customers, Products and Countries
print("\nUNIQUE VALUES")
print("=" * 50)

print("Unique Customers:", df["CustomerID"].nunique())
print("Unique Products:", df["StockCode"].nunique())
print("Unique Countries:", df["Country"].nunique())
print("Unique Invoices:", df["InvoiceNo"].nunique())


# 7. Check Negative and Zero Values
print("\nNEGATIVE AND ZERO VALUES")
print("=" * 50)

print("Negative Quantity:", (df["Quantity"] < 0).sum())
print("Zero Quantity:", (df["Quantity"] == 0).sum())
print("Negative Unit Price:", (df["UnitPrice"] < 0).sum())
print("Zero Unit Price:", (df["UnitPrice"] == 0).sum())

print("\nSTEP 2 COMPLETED SUCCESSFULLY!")
```

```python

# ==========================================
# STEP 3: DATA CLEANING
# ==========================================

# 1. Create a copy of original dataset

df_clean = df.copy()

print("ORIGINAL DATA SHAPE:", df_clean.shape)


# 2. Remove duplicate records

df_clean = df_clean.drop_duplicates()

print("After removing duplicates:", df_clean.shape)


# 3. Check missing descriptions

print("\nMissing Descriptions:",
      df_clean["Description"].isnull().sum())

# Create dataset for product and sales analysis
df_clean = df_clean.dropna(subset=["Description"])


# 4. Identify missing Customer IDs

print("\nMissing Customer IDs:",
      df_clean["CustomerID"].isnull().sum())

# We keep these records for general sales analysis.
# They will be excluded only from customer segmentation.


# 5. Identify cancelled and returned transactions

cancelled = df_clean["InvoiceNo"].astype(str).str.startswith("C")

df_returns = df_clean[
    (df_clean["Quantity"] < 0) | cancelled
].copy()

print("\nReturned/Cancelled Records:",
      df_returns.shape[0])


# 6. Identify negative prices

df_invalid_price = df_clean[
    df_clean["UnitPrice"] < 0
].copy()

print("Records with negative prices:",
      df_invalid_price.shape[0])


# 7. Create clean sales dataset

df_sales = df_clean[
    (df_clean["Quantity"] > 0) &
    (df_clean["UnitPrice"] > 0) &
    (~cancelled)
].copy()


# 8. Final dataset information

print("\nCLEANING RESULTS")
print("=" * 50)

print("Original Records:", df.shape[0])
print("Cleaned Records:", df_clean.shape[0])
print("Valid Sales Records:", df_sales.shape[0])
print("Return Records:", df_returns.shape[0])

print("\nRemaining Duplicates:",
      df_clean.duplicated().sum())

print("\nSTEP 3 COMPLETED SUCCESSFULLY!")

display(df_sales.head())
```

```python

# ==========================================
# STEP 4: FEATURE ENGINEERING
# ==========================================

# 1. Create Revenue column

df_sales["Revenue"] = (
    df_sales["Quantity"] * df_sales["UnitPrice"]
)

# 2. Extract Year

df_sales["Year"] = df_sales["InvoiceDate"].dt.year

# 3. Extract Month

df_sales["Month"] = df_sales["InvoiceDate"].dt.month

# 4. Extract Month Name

df_sales["MonthName"] = (
    df_sales["InvoiceDate"].dt.month_name()
)

# 5. Extract Day

df_sales["Day"] = df_sales["InvoiceDate"].dt.day

# 6. Extract Day Name

df_sales["DayName"] = (
    df_sales["InvoiceDate"].dt.day_name()
)

# 7. Extract Hour

df_sales["Hour"] = df_sales["InvoiceDate"].dt.hour


# ==========================================
# CHECK RESULTS
# ==========================================

print("FEATURE ENGINEERING COMPLETED!")
print("=" * 50)

print("Updated Dataset Shape:", df_sales.shape)

print("\nNew Columns:")
print(df_sales.columns.tolist())

print("\nFirst 10 Records:")
display(df_sales.head(10))

print("\nRevenue Statistics:")
display(df_sales["Revenue"].describe())
```

```python

# ==========================================
# STEP 5: SALES ANALYSIS
# ==========================================

# 1. Total Revenue

total_revenue = df_sales["Revenue"].sum()

print("TOTAL REVENUE: £", round(total_revenue, 2))


# 2. Total Quantity Sold

total_quantity = df_sales["Quantity"].sum()

print("\nTOTAL QUANTITY SOLD:", total_quantity)


# 3. Total Unique Orders

total_orders = df_sales["InvoiceNo"].nunique()

print("\nTOTAL UNIQUE ORDERS:", total_orders)


# 4. Total Unique Customers

total_customers = df_sales["CustomerID"].nunique()

print("\nTOTAL UNIQUE CUSTOMERS:", total_customers)


# 5. Average Order Value

order_revenue = df_sales.groupby("InvoiceNo")["Revenue"].sum()

average_order_value = order_revenue.mean()

print("\nAVERAGE ORDER VALUE: £",
      round(average_order_value, 2))


# 6. Top 10 Best-Selling Products by Quantity

top_products = (
    df_sales.groupby("Description")["Quantity"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("\nTOP 10 BEST-SELLING PRODUCTS")
display(top_products)


# 7. Top 10 Products by Revenue

top_revenue_products = (
    df_sales.groupby("Description")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("\nTOP 10 PRODUCTS BY REVENUE")
display(top_revenue_products)


print("\nSTEP 5 COMPLETED SUCCESSFULLY!")
```

```python

# ==========================================
# STEP 7: COMPLETE RFM CUSTOMER ANALYSIS
# ==========================================

# 1. Prepare customer data
customer_df = df_sales.dropna(subset=["CustomerID"]).copy()
customer_df["CustomerID"] = customer_df["CustomerID"].astype(int)

latest_date = customer_df["InvoiceDate"].max()
analysis_date = latest_date + pd.Timedelta(days=1)

# 2. Calculate RFM values
rfm = customer_df.groupby("CustomerID").agg({
    "InvoiceDate": lambda x: (analysis_date - x.max()).days,
    "InvoiceNo": "nunique",
    "Revenue": "sum"
})

rfm.rename(columns={
    "InvoiceDate": "Recency",
    "InvoiceNo": "Frequency",
    "Revenue": "Monetary"
}, inplace=True)

# 3. Assign RFM scores
rfm["R_Score"] = pd.qcut(
    rfm["Recency"].rank(method="first"),
    5, labels=[5, 4, 3, 2, 1]
).astype(int)

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    5, labels=[1, 2, 3, 4, 5]
).astype(int)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    5, labels=[1, 2, 3, 4, 5]
).astype(int)

rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str) +
    rfm["F_Score"].astype(str) +
    rfm["M_Score"].astype(str)
)

# 4. Create customer segments
def segment_customer(row):
    if row["R_Score"] >= 4 and row["F_Score"] >= 4 and row["M_Score"] >= 4:
        return "Champions"
    elif row["R_Score"] >= 3 and row["F_Score"] >= 3:
        return "Loyal Customers"
    elif row["R_Score"] >= 4:
        return "Recent Customers"
    elif row["R_Score"] <= 2 and row["F_Score"] >= 3:
        return "At Risk"
    else:
        return "Regular Customers"

rfm["Customer_Segment"] = rfm.apply(segment_customer, axis=1)

# 5. Display RFM results
print("========== RFM CUSTOMER ANALYSIS ==========")
print("Total Customers:", len(rfm))
print("Analysis Date:", analysis_date)

print("\nRFM SUMMARY:")
display(rfm[["Recency", "Frequency", "Monetary"]].describe().round(2))

print("\nCUSTOMER SEGMENT COUNTS:")
display(rfm["Customer_Segment"].value_counts().to_frame("Number of Customers"))

print("\nSAMPLE CUSTOMER DATA:")
display(rfm.head(10))

# 6. Visualize customer segments
plt.figure(figsize=(10, 6))

sns.countplot(
    data=rfm,
    y="Customer_Segment",
    order=rfm["Customer_Segment"].value_counts().index
)

plt.title("Customer Segmentation Using RFM", fontsize=16)
plt.xlabel("Number of Customers")
plt.ylabel("Customer Segment")
plt.tight_layout()
plt.show()

print("\nSTEP 7 COMPLETED SUCCESSFULLY!")
```

```python

# ==========================================
# STEP 8: K-MEANS CUSTOMER CLUSTERING
# ==========================================

# 1. Import required libraries
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import numpy as np

# 2. Select RFM features
X = rfm[["Recency", "Frequency", "Monetary"]].copy()

# 3. Handle skewness using log transformation
X_log = np.log1p(X)

# 4. Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_log)

# 5. Find the best number of clusters using Elbow Method
inertia = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

# 6. Display Elbow Graph
plt.figure(figsize=(9, 5))
plt.plot(range(2, 11), inertia, marker="o")
plt.title("Elbow Method for Optimal Clusters")
plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")
plt.grid(True)
plt.tight_layout()
plt.show()

# 7. Train final K-Means model
# Start with 4 clusters as a practical initial choice
kmeans_final = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

rfm["Cluster"] = kmeans_final.fit_predict(X_scaled)

# 8. Evaluate clustering
silhouette = silhouette_score(X_scaled, rfm["Cluster"])

print("========== K-MEANS RESULTS ==========")
print("Number of Clusters:", rfm["Cluster"].nunique())
print("Silhouette Score:", round(silhouette, 3))

# 9. Analyze customer clusters
cluster_summary = rfm.groupby("Cluster").agg({
    "Recency": "mean",
    "Frequency": "mean",
    "Monetary": "mean",
    "Customer_Segment": "count"
}).round(2)

cluster_summary.rename(
    columns={"Customer_Segment": "Customer_Count"},
    inplace=True
)

print("\nCLUSTER SUMMARY:")
display(cluster_summary)

# 10. Visualize customer clusters
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=rfm,
    x="Recency",
    y="Monetary",
    hue="Cluster",
    palette="viridis",
    s=60
)

plt.title("Customer Clusters - K-Means")
plt.xlabel("Recency (Days)")
plt.ylabel("Monetary Value (£)")
plt.tight_layout()
plt.show()

print("\nSTEP 8 COMPLETED SUCCESSFULLY!")
```

```python

# ==========================================
# STEP 9: BUSINESS INSIGHTS & RECOMMENDATIONS
# ==========================================

# 1. Overall business performance
total_revenue = df_sales["Revenue"].sum()
total_orders = df_sales["InvoiceNo"].nunique()
total_customers = rfm.shape[0]

average_order_value = (
    df_sales.groupby("InvoiceNo")["Revenue"].sum().mean()
)

# 2. Best-selling product
best_product = (
    df_sales.groupby("Description")["Quantity"]
    .sum()
    .idxmax()
)

# 3. Highest revenue country
best_country = (
    df_sales.groupby("Country")["Revenue"]
    .sum()
    .idxmax()
)

# 4. Best customer segment
best_segment = rfm["Customer_Segment"].value_counts().idxmax()

# 5. Highest revenue customer
top_customer = rfm["Monetary"].idxmax()
top_customer_spending = rfm["Monetary"].max()

# 6. Cluster with highest average spending
best_cluster = (
    rfm.groupby("Cluster")["Monetary"]
    .mean()
    .idxmax()
)

# 7. Display business insights
print("========== BUSINESS PERFORMANCE ==========")
print(f"Total Revenue: £{total_revenue:,.2f}")
print(f"Total Orders: {total_orders:,}")
print(f"Total Customers: {total_customers:,}")
print(f"Average Order Value: £{average_order_value:,.2f}")

print("\n========== KEY BUSINESS INSIGHTS ==========")
print("Best-selling Product:", best_product)
print("Highest Revenue Country:", best_country)
print("Largest Customer Segment:", best_segment)
print("Highest Spending Customer ID:", top_customer)
print(f"Highest Customer Spending: £{top_customer_spending:,.2f}")
print("Highest Average-Spending Cluster:", best_cluster)

print("\n========== BUSINESS RECOMMENDATIONS ==========")
print("1. Promote best-selling products through targeted campaigns.")
print("2. Focus marketing efforts on high-revenue countries.")
print("3. Reward loyal and champion customers with exclusive offers.")
print("4. Re-engage at-risk customers through personalized campaigns.")
print("5. Study high-spending clusters to identify valuable customer behaviour.")

print("\nSTEP 9 COMPLETED SUCCESSFULLY!")
```
